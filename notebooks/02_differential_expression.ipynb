# Stage 2 — Differential Expression Analysis

This notebook compares gene expression between:

- Sensitive samples: TS1, TS2, TS3
- Resistant samples: TR1, TR2, TR3

The goal is to identify genes whose expression differs between
the Resistant and Sensitive groups.

The validated gene-level count matrix from Stage 1 is used as
the starting point.

## Analysis workflow

1. Load validated gene counts
2. Load sample metadata
3. Validate the input data
4. Filter genes with insufficient counts
5. Prepare the count matrix for PyDESeq2
6. Run PyDESeq2
7. Perform the Resistant vs Sensitive contrast
8. Examine differential-expression results
9. Save the results for downstream pathway and CMap analysis
