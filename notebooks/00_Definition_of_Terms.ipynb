# Definition of Terms

| Term                         | Simple definition                                                                                                               | Easy way to think about it                                                                 |
| ---------------------------- | ------------------------------------------------------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------ |
| **RNA-seq**                  | A method for measuring which genes are active in a cell and roughly how much.                                                   | Taking a **snapshot of which genes are "turned on"**                                       |
| **RNA**                      | A molecule that carries instructions copied from DNA and helps cells make proteins.                                             | A **working copy of instructions** from DNA                                                |
| **Gene**                     | A section of DNA containing instructions for something the cell does, often making a protein.                                   | A **recipe in a huge cookbook**                                                            |
| **Transcript**               | An RNA copy made from a gene. One gene can produce multiple different transcripts.                                              | A **specific version of a recipe**                                                         |
| **Salmon**                   | A computer program that estimates how much of each RNA transcript is present in an RNA-seq sample.                              | A **counter that estimates how much of each recipe version was used**                      |
| **`quant.sf`**               | A file produced by Salmon containing transcript-level measurements.                                                             | Salmon's **report card**                                                                   |
| **`NumReads`**               | Salmon's estimate of how many sequencing reads came from a particular transcript.                                               | An estimate of **how many pieces of evidence** point to that transcript                    |
| **TPM**                      | A measurement of transcript abundance that accounts for transcript length and the total amount of RNA measured.                 | A way of saying **"how much of this transcript is present compared with everything else"** |
| **Ensembl**                  | A database containing information about genes, transcripts, and genomes.                                                        | A huge **gene encyclopedia/database**                                                      |
| **Ensembl ID**               | A unique ID assigned to a gene or transcript.                                                                                   | Like a **barcode** for a gene or transcript                                                |
| **`ENSG...`**                | An Ensembl **gene** ID.                                                                                                         | `ENSG...` = **gene barcode**                                                               |
| **`ENST...`**                | An Ensembl **transcript** ID.                                                                                                   | `ENST...` = **transcript barcode**                                                         |
| **`tx2gene`**                | A table that tells us which gene each transcript belongs to.                                                                    | A **translation table: transcript → gene**                                                 |
| **Mapping**                  | Connecting one ID to another using a reference table.                                                                           | Matching **student ID → student name**                                                     |
| **Gene-level counts**        | Transcript measurements combined together so we have one value for each gene.                                                   | Adding up all the **versions of a recipe** to get one total for the recipe                 |
| **Count matrix**             | A table where rows are genes, columns are samples, and the values are expression measurements.                                  | A big **spreadsheet of gene activity**                                                     |
| **Sample**                   | One biological specimen being analyzed.                                                                                         | One **individual experiment/specimen**                                                     |
| **Biological replicate**     | Different samples representing the same biological condition.                                                                   | Three separate people from the **same group**                                              |
| **Sensitive (TS)**           | In your project, the samples classified as treatment-sensitive.                                                                 | One **group being studied**                                                                |
| **Resistant (TR)**           | In your project, the samples classified as treatment-resistant.                                                                 | The **other group being studied**                                                          |
| **QC**                       | **Quality Control** — checks that your data looks reliable before analysis.                                                     | **Checking your work before submitting it**                                                |
| **Input validation**         | Checking that the expected files exist and have the expected structure.                                                         | Making sure you have **all the correct ingredients**                                       |
| **Transcript ID validation** | Checking that the transcript IDs in your Salmon files look correct.                                                             | Checking that all the **barcodes are readable**                                            |
| **Mapping validation**       | Checking that Salmon transcripts can be connected to Ensembl genes.                                                             | Making sure the **translation table actually works**                                       |
| **Match rate**               | The percentage of Salmon transcripts successfully found in the mapping table.                                                   | "What percentage of items could we **successfully match**?"                                |
| **Mapping uniqueness**       | Checking that one transcript doesn't accidentally map to multiple genes.                                                        | Making sure one **barcode doesn't point to two different products**                        |
| **Library size**             | The total number of estimated reads/counts in a sample.                                                                         | How much **sequencing information** we got from that sample                                |
| **Sample correlation**       | A number describing how similar two samples' gene-expression patterns are.                                                      | "How similarly do these two samples **behave**?"                                           |
| **Pearson correlation**      | A statistical measure ranging from -1 to +1 describing how strongly two sets of values move together.                           | A **similarity score** for two patterns                                                    |
| **PCA**                      | **Principal Component Analysis** — a way to visualize which samples are similar or different across thousands of genes at once. | Turning a **huge spreadsheet into a simple map**                                           |
| **Outlier**                  | A sample that behaves very differently from the others.                                                                         | The **odd one out**                                                                        |
| **Differential expression**  | Finding genes whose expression differs between two groups.                                                                      | Asking **"Which genes are more or less active in Group A vs Group B?"**                    |
| **DEG**                      | **Differentially Expressed Gene** — a gene whose expression shows a statistically significant difference between groups.        | A gene that **stands out between the two groups**                                          |
| **PyDESeq2**                 | A Python tool used to statistically analyze gene-expression differences between groups.                                         | The **statistical referee** deciding which differences are convincing                      |
| **Normalization**            | Adjusting measurements so samples can be fairly compared despite differences in sequencing depth.                               | Making sure we're comparing **apples to apples**                                           |
| **Statistical significance** | A measure of how unlikely an observed difference would be if there were actually no real difference.                            | "Is this difference **probably real or could it just be chance?**"                         |
| **p-value**                  | A number used to assess evidence against the idea that there is no difference.                                                  | A measure related to **how surprising the result would be by chance**                      |
| **FDR**                      | **False Discovery Rate** — helps control the number of genes incorrectly called significant when testing thousands of genes.    | A **safety check against too many false alarms**                                           |
| **log2 fold change**         | Measures how much gene expression changes between two groups on a logarithmic scale.                                            | How many times **higher or lower** the expression is                                       |
| **Upregulated**              | A gene has higher expression in one group compared with another.                                                                | The gene is **more active**                                                                |
| **Downregulated**            | A gene has lower expression in one group compared with another.                                                                 | The gene is **less active**                                                                |
| **Reference genome**         | A standard DNA sequence used as a reference when analyzing genetic data.                                                        | A **master copy of the instruction book**                                                  |
| **GRCh38**                   | The human reference genome assembly we're using.                                                                                | The specific **version of the human DNA reference book**                                   |
| **Genome assembly**          | A particular organized version of the reference genome.                                                                         | A specific **edition of the DNA map**                                                      |
| **`ENST000... .1/.2`**       | A transcript ID with a version number attached.                                                                                 | The transcript's **barcode + edition/version number**                                      |
| **`.1`, `.2`, etc.**         | The version number attached to an Ensembl ID.                                                                                   | Like **Version 1, Version 2** of the same record                                           |
